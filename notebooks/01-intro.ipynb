```python
import numpy as np
import matplotlib.pyplot as plt
from examples.generate_synthetic import generate
from bias_corrector.core import quantile_mapping_fit, quantile_mapping_apply, train_nn_correction, apply_nn_correction

t, obs, sim = generate(1000, seed=0)
plt.figure(figsize=(10,3))
plt.plot(t[:200], obs[:200], label='obs')
plt.plot(t[:200], sim[:200], label='sim')
plt.legend()
plt.title('Synthetic observed vs simulated (first 200 samples)')
plt.show()
```

## Quantile mapping
Fit a quantile mapping between `sim` and `obs`, then apply it.

```python
sim_q, obs_q = quantile_mapping_fit(obs, sim, q=1000)
sim_q[:5], obs_q[:5]
```

```python
corrected_qm = quantile_mapping_apply(sim, sim_q, obs_q)
plt.figure(figsize=(10,3))
plt.plot(t[:200], obs[:200], label='obs')
plt.plot(t[:200], sim[:200], label='sim')
plt.plot(t[:200], corrected_qm[:200], label='corrected (QM)')
plt.legend()
plt.title('Quantile-mapping correction (first 200 samples)')
plt.show()
```

## Neural residual correction
Train a small MLP to predict the residual `obs - sim` from `sim` and apply it.

```python
model = train_nn_correction(sim, obs, hidden_layer_sizes=(50,))
corrected_nn = apply_nn_correction(sim, model)
plt.figure(figsize=(10,3))
plt.plot(t[:200], obs[:200], label='obs')
plt.plot(t[:200], sim[:200], label='sim')
plt.plot(t[:200], corrected_nn[:200], label='corrected (NN)')
plt.legend()
plt.title('NN residual correction (first 200 samples)')
plt.show()
```
